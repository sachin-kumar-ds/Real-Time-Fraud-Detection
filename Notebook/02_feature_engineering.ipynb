# Feature Engineering

```python
import pandas as pd
import numpy as np

df = pd.read_csv("C:/Users/LENOVO/OneDrive/Desktop/Real Time Fraud Detection/Dataset/Raw/bs140513_032310.csv")

# Drop Useless Columns

df = df.drop(columns=["zipcodeOri","zipMerchant"])

# Remove quotes from cateogiral Values

catergorial_cols = [
    "customer",
    "age",
    "gender",
    "merchant",
    "category"
]

for col in catergorial_cols:
    df[col] = df[col].str.replace("'", "")


df["log_amount"] = np.log1p(df["amount"])

# Creating high Flag Amount

threshold = df["amount"].quantile(0.95)

df["high_amount_flag"] = (
    df["amount"] > threshold
).astype(int)

# Category Fraud Rate Feature
# One of Strongest Feature

category_fraud_rate = (
    df.groupby("category")['fraud']
    .mean()
)

df["category_fraud_rate"] = (
    df["category"].map(category_fraud_rate)
)

# Merchant Fraud Rate Feature

merchant_fraud_rate = (
    df.groupby("merchant")["fraud"]
    .mean()
)

df["merchant_fraud_rate"] = (
    df["merchant"]
    .map(merchant_fraud_rate)
)

# Customer Transaction Count

customer_tx_count = (
    df.groupby("customer")["amount"]
    .transform("count")
)

df["customer_tx_count"] = customer_tx_count

# Customer Average Amount

customer_avg_amount = (
    df.groupby("customer")["amount"]
    .transform("mean")
)

df["customer_avg_amount"] = customer_avg_amount

# Amount Deviation Feature
# Very Important Feature

df["amount_deviation"] = (
    df["amount"] - df["customer_avg_amount"]
)

customer_fraud_history = (
    df.groupby("customer")["fraud"]
    .transform("mean")
)

df["customer_fraud_history"] = (customer_fraud_history)



```
