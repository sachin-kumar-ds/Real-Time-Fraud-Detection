```python
import pandas as pd

df = pd.read_csv("C:/Users/LENOVO/OneDrive/Desktop/Real Time Fraud Detection/Dataset/Raw/bs140513_032310.csv")

print(df.head())

print("\n Shape:")
print(df.shape)

print("\n Columns: ")
print(df.columns)

print("\n Data Types: ")
print(df.dtypes)

print("\n Missing Values: ")
print(df.isnull().sum())

print("\n Duplicate Rows: ")
print(df.duplicated().sum())

print("\n Fraud Distribuion: ")
print(df["fraud"].value_counts())

print("\n Fraud Percentage: ")
print(df["fraud"].value_counts(normalize=True)*100)
# The normalized = True Normalize the output into 0 and 1 
# Like 0.8 and 0.2 and multiply by 100 to get the percentage
```
